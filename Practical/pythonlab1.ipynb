Program 1: Variables and Data Types

Write a Python program to declare variables of type int, float, string, and boolean, and display their values along with their data types.
Write a program to check the data type of different variables using the type() function.
Modify the values of the variables and observe how their data types change.
Create your own variables representing a student's details and print their values and data types.

```python
# Program 1: Variables and Data Types

# Step 1: Declare variables of different data types
num = 10                 # int
price = 99.99            # float
name = "Sita"        # string
is_student = True        # boolean

# Display values and data types
print("Initial Variables:")
print("num =", num, "Data Type:", type(num))
print("price =", price, "Data Type:", type(price))
print("name =", name, "Data Type:", type(name))
print("is_student =", is_student, "Data Type:", type(is_student))

# Step 2: Check data types using type()
print("\nChecking Data Types:")
print(type(num))
print(type(price))
print(type(name))
print(type(is_student))

# Step 3: Modify the values and observe data type changes
num = 25.5
price = "One Hundred"
name = False
is_student = 1

print("\nAfter Modifying Variables:")
print("num =", num, "Data Type:", type(num))
print("price =", price, "Data Type:", type(price))
print("name =", name, "Data Type:", type(name))
print("is_student =", is_student, "Data Type:", type(is_student))

# Step 4: Student details
student_name = "Sita"
student_age = 20
student_cgpa = 9.5
student_branch = "CSE"
student_hosteller = True

print("\nStudent Details:")
print("Name =", student_name, "| Data Type:", type(student_name))
print("Age =", student_age, "| Data Type:", type(student_age))
print("CGPA =", student_cgpa, "| Data Type:", type(student_cgpa))
print("Branch =", student_branch, "| Data Type:", type(student_branch))
print("Hosteller =", student_hosteller, "| Data Type:", type(student_hosteller))
```

Program 2: Operators

Write a Python program to demonstrate arithmetic operators.
Write a program to demonstrate comparison operators using two numbers.
Write a program to demonstrate logical operators (and, or, not).
Write a program to determine whether a student is eligible for placement based on CGPA and attendance using logical operators.

```python

# Program 2: Operators

# Step 1: Arithmetic Operators
a = 20
b = 5

print("Arithmetic Operators:")
print("a =", a, "b =", b)
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor Division:", a // b)
print("Modulus:", a % b)
print("Exponent:", a ** b)

# Step 2: Comparison Operators
print("\nComparison Operators:")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)

# Step 3: Logical Operators
x = True
y = False

print("\nLogical Operators:")
print("x and y:", x and y)
print("x or y:", x or y)
print("not x:", not x)
print("not y:", not y)

# Step 4: Student Placement Eligibility
cgpa = 8.2
attendance = 80

eligible = (cgpa >= 7.0) and (attendance >= 75)

print("\nStudent Placement Eligibility:")
print("CGPA:", cgpa)
print("Attendance:", attendance, "%")

if eligible:
    print("Result: Student is Eligible for Placement.")
else:
    print("Result: Student is Not Eligible for Placement.")
```

Program 5: Dictionaries

Write a Python program to create a dictionary and display all its key-value pairs.
Write a program to access the value associated with a specific key in a dictionary.
Write a program to iterate through a dictionary using the items() method.
Write a program to store different machine learning models in a dictionary and display their names and objects.

A dictionary gives a key:value pairs in curly braces

```python
from sklearn.linear_model import Ridge,Lasso,ElasticNet

models ={
    "Ridge": Ridge(alpha=1.0),
    "Lasso":Lasso(alpha=0.01),
    "ElasticNet":ElasticNet(alpha=0.01),
}

print(models) #the whole dictionary
print(models["Ridge"]) #access one value by its key

#Looping over a dictionary's key and values together

for name, model in models.items():
  print(f"Model name: {name} -> Object: {model}")
```

```python
# Program 5: Dictionaries

# Step 1: Create a dictionary
student = {
    "Name": "Rahul",
    "Age": 20,
    "Branch": "CSE",
    "CGPA": 8.5
}

# Display all key-value pairs
print("Student Dictionary:")
for key, value in student.items():
    print(key, ":", value)

# Step 2: Access the value of a specific key
print("\nAccessing a Specific Key:")
print("Name:", student["Name"])
print("CGPA:", student["CGPA"])

# Step 3: Iterate through the dictionary using items()
print("\nIterating Through Dictionary:")
for key, value in student.items():
    print(key, "->", value)

# Step 4: Store machine learning models in a dictionary
ml_models = {
    "Linear Regression": "LinearRegression()",
    "Decision Tree": "DecisionTreeClassifier()",
    "Random Forest": "RandomForestClassifier()",
    "K-Nearest Neighbors": "KNeighborsClassifier()",
    "Support Vector Machine": "SVC()"
}

print("\nMachine Learning Models:")
for model, obj in ml_models.items():
    print(model, ":", obj)
```

Program 6: Conditional Statements

Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.

```python
# Program 6: Conditional Statements

# Step 1: Classify a student's CGPA
cgpa = 8.2

print("CGPA Classification:")
if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")

# Step 2: Check for missing values in a dataset
dataset = [8.2, 85, None, 90]

print("\nChecking for Missing Values:")
if None in dataset:
    print("Missing values are present in the dataset.")
else:
    print("No missing values found.")

# Step 3: Check placement eligibility based on CGPA
print("\nPlacement Eligibility:")
if cgpa >= 7.0:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")

# Step 4: Assign grades based on marks
marks = 82

print("\nGrade Assignment:")
if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```

Program 7: Loops

Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.

```python
# Program 7: Loops

# Step 1: Create a list
scores = [78, 85, 92, 67, 88, 95, 73]

# Display all elements using a for loop
print("List Elements:")
for score in scores:
    print(score)

# Step 2: Calculate the average of the list
total = 0
for score in scores:
    total += score

average = total / len(scores)
print("\nAverage:", average)

# Step 3: Display values greater than a given threshold
threshold = 80

print("\nValues Greater Than", threshold, ":")
for score in scores:
    if score > threshold:
        print(score)

# Step 4: Find the maximum and minimum values using loops
maximum = scores[0]
minimum = scores[0]

for score in scores:
    if score > maximum:
        maximum = score
    if score < minimum:
        minimum = score

print("\nMaximum Value:", maximum)
print("Minimum Value:", minimum)

# Step 5: Count the number of elements satisfying a condition
count = 0

for score in scores:
    if score >= 75:
        count += 1

print("\nNumber of Scores Greater Than or Equal to 75:", count)
```
