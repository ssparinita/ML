Program 1: Variables and Data Types
Write a Python program to declare variables of type int, float, string, and boolean, and display their values along with their data types.

Write a program to check the data type of different variables using the type() function.

Modify the values of the variables and observe how their data types change.

Create your own variables representing a student's details and print their values and data types.

```python
# 1. Declare variables of different types and display their values and data types
a = 10        # int
b = 5.5       # float
c = "Hello"   # string
d = True      # boolean

print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))

# 2. Check the data type using the type() function
x = 100
y = "Python"
print(type(x))
print(type(y))

# 3. Modify the values of variables and observe how their data types change
var = 10        # Initially an int
print(var, type(var))

var = 10.5      # Changed to a float
print(var, type(var))

var = "Done"    # Changed to a string
print(var, type(var))

# 4. Create variables representing a student's details and print their values and data types
student_name = "Alice"
student_age = 20
student_cgpa = 8.8
is_enrolled = True

print(student_name, type(student_name))
print(student_age, type(student_age))
print(student_cgpa, type(student_cgpa))
print(is_enrolled, type(is_enrolled))
```

Program 2: Operators
Write a Python program to demonstrate arithmetic operators.
Write a program to demonstrate comparison operators using two numbers.
Write a program to demonstrate logical operators (and, or, not).
Write a program to determine whether a student is eligible for placement based on CGPA and attendance using logical operators.

```python
# Arithmetic & Comparison Operators
a = 7
b = 2
print("a+b=", a+b)
print("a-b=", a-b)
print("a*b=", a*b)
print("a/b=", a/b)
print("a**2=", a**2)

cgpa = 9.7
print("cgpa = 9.7 :",cgpa == 9.7)
print("cgpa > 9   :",cgpa>9)
print("cgpa < 6   :",cgpa<6)

attendance = 85
placed = (cgpa > 9) and (attendance > 80)
print("placed : ",placed)
```

```python
#STRINGS AND F-STRINGS

model_name = "Logistic Regression"
accuracy = 0.87345

# Old, clunky way (avoid this):
print("Accuracy of " + model_name + " is " + str(accuracy))

# f-string way (use this):
print(f"Accuracy of {model_name} is {accuracy}")

# ':.3f' inside the braces rounds a float to 3 decimal places -
# you will use this in almost every experiment to print clean metrics.
print(f"Accuracy of {model_name} is {accuracy:.3f}")
```
